### 2.2 Renombrar columnas

```python
# Estandarizar los nombres de las columnas de eco
eco_df = eco_df.rename(columns={
    'Year': 'year',
    'City': 'city',
    'Country': 'country',
    'City GDP/capita': 'city_gdp_capita',
    'Unemployment %': 'unemployment',
    'PM2.5 (μg/m³)': 'pm2_5',
    'Population (M)': 'population_m'
})
# verificar cambios
eco_df.columns
```


### 2.3 Corregir formatos numéricos y de fecha

**🎯Objetivo:**
Asegurar que las columnas de fechas y valores numéricos estén en formatos correctos para permitir análisis, cálculos y comparaciones precisas.

**Instrucciones:**

- Convierte las columnas de fecha de `traffic` a formato `datetime`. Haz el cambio a prueba de errores.
- En el dataset `eco`, limpia los valores numéricos:
    - En `city_gdp_capita`: elimina separadores de miles (`.`) y reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
    - En `unemployment_pct`: elimina el símbolo de porcentaje (`%`) y reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
    - En `population_m`: reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
- Finalmente, crea una nueva columna llamada `population` multiplicando `population_m` por 1,000,000 para obtener la población total.


```python
# Convertir las columnas de traffic a tipo fecha con pd.to_datetime()

# Convertir ambas columnas a formato datetime
date_cols = ['update_time_utc', 'update_time_utc_week_ago']

for col in date_cols:
    traffic_df[col] = pd.to_datetime(traffic_df[col], errors='coerce')
# verificar el cambio
traffic_df.info()
```

```python
# city_gdp_capita: eliminar separador de miles (.) y reemplazar coma decimal (,) por punto (.)
eco_df['city_gdp_capita'] = (
    eco_df['city_gdp_capita']
    .astype(str)
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
    .str.strip()
)
eco_df['city_gdp_capita'] = pd.to_numeric(eco_df['city_gdp_capita'], errors='coerce')

# unemployment_pct: eliminar (%) y reemplazar coma decimal (,) por punto (.)
eco_df['unemployment'] = (
    eco_df['unemployment']
    .astype(str)
    .str.replace('%', '', regex=False)
    .str.replace(',', '.', regex=False)
    .str.strip()
)
eco_df['unemployment'] = pd.to_numeric(eco_df['unemployment'], errors='coerce')

# population_m: reemplazar coma decimal (,) por punto (.)
eco_df['population_m'] = (
    eco_df['population_m']
    .astype(str)
    .str.replace(',', '.', regex=False)
    .str.strip()
)
eco_df['population_m'] = pd.to_numeric(eco_df['population_m'], errors='coerce')

# -------------------------------------------------------------
# Crear columna population (población total)
# -------------------------------------------------------------
eco_df['population'] = eco_df['population_m'] * 1_000_000

# verificar el cambio
eco_df.info()
eco_df.head(3)
```

## 🧩Paso 4: Analizar y resumir datos de movilidad

Como el dataset de tráfico contiene **múltiples registros por ciudad**. En esta parte, calcularás los promedios anuales por ciudad para simplificar el análisis y obtener una visión más clara de las tendencias generales.

### 4.1 Calcular promedios de tráfico por ciudad

**🎯Objetivo:**
Obtener una vista consolidada del tráfico promedio por ciudad y año, para analizar patrones generales sin depender de datos diarios.

**Instrucciones**

- Agrupa los datos por `city`, `country` y `year`.
- Calcula el promedio **solo de las métricas de tráfico más relevantes**: como `jams_delay`, `traffic_index_live`, `jams_length_kms`, `jams_count`, `mins_delay`, y tiempos de viaje (`travel_time_live_per_10kms_mins` y `travel_time_hist_per_10kms_mins`).
- Guarda el resultado como `traffic_city_year_2024`, mantén las columnas como variables (no índices).


```python
# Lista de métricas de tráfico con los nombres exactos de las columnas
traffic_metrics = [
    'jams_delay',
    'traffic_index_live',
    'jams_length_in_kms',
    'jams_count',
    'mins_delay',
    'travel_time_live_per_10_kms_mins',
    'travel_time_historic_per_10_kms_mins'
]

# Calcular los promedios de tráfico por ciudad, país y año
traffic_city_year_2024 = (
    traffic_2024
    .groupby(['city', 'country', 'year'])[traffic_metrics]
    .mean()
    .reset_index()
)

# Mostrar resultado
traffic_city_year_2024.head()
```

## 🧩Paso 5: Unir movilidad y economía

Combinar datasets te permite analizar cómo se relacionan los indicadores económicos con los de movilidad.

### 5.1 Unir tráfico (tabla principal) con indicadores económicos

**🎯Objetivo:**
Combinar la información de tráfico y economía en un solo DataFrame para analizar cómo las condiciones económicas se relacionan con la movilidad urbana.

**Instrucciones**
- Selecciona solo las **columnas relevantes** de cada dataset (por ejemplo, variables clave de tráfico y de economía).
- Usa `.copy()` al crear subconjuntos para evitar modificar el dataset original.
- Une ambos DataFrames y define como **claves de unión** a `city` y `year`.
- Mantén solo las ciudades y años presentes en ambos datasets.
- Guarda el resultado en una nueva variable llamada `merged` y muestra las primeras 5 filas.


```python
# 1. Definir columnas clave 
left_cols = [
    col for col in [
        'city', 'country', 'year', 'jams_delay', 'traffic_index_live',
        'jams_length_kms', 'jams_length_in_kms', 'jams_count', 'mins_delay',
        'travel_time_live_per_10kms_mins', 'travel_time_live_per_10_kms_mins',
        'travel_time_hist_per_10kms_mins', 'travel_time_historic_per_10_kms_mins'
    ] if col in traffic_city_year_2024.columns
]

right_cols = [
    col for col in [
        'city', 'year', 'city_gdp_capita', 'unemployment_pct', 'unemployment',
        'pm25', 'pm2_5', 'population'
    ] if col in eco_2024.columns
]

# 2. Usar .copy() para crear los subconjuntos reducidos
traffic_2024_small = traffic_city_year_2024[left_cols].copy()
eco_2024_small = eco_2024[right_cols].copy()

# 3. Unir datasets con inner join en 'city' y 'year'
merged = traffic_2024_small.merge(eco_2024_small, on=['city', 'year'], how='inner')

# 4. Mostrar las primeras 5 filas
merged.head()
```

## 🧩Paso 6: Visualización y análisis de relaciones

Ahora que tienes un dataset limpio y unificado, es momento de **visualizar patrones**.
Los gráficos te ayudarán a entender cómo se relacionan las variables económicas con las de movilidad urbana.

### 6.1 Visualizar relaciones entre economía y tráfico

**🎯Objetivo:**
Analizar visualmente la distribución y la relación entre indicadores de tráfico y economía en 2024, para identificar posibles patrones o tendencias generales entre ambas variables.

**Instrucciones**
- Usa las librerías `seaborn` y `matplotlib.pyplot` para generar los gráficos.
- Visualiza la distribución del **tráfico** (`jams_delay`) mediante:
    - **Boxplot** → para observar la media, mediana y detectar valores atípicos.
- Visualiza la distribución de la **economía** (`city_gdp_capita`) mediante:
    - **Histograma** → para analizar la forma de la distribución y el valor promedio del PIB per cápita.
- Finalmente, **compara ambas variables**, para observar si existe alguna relación entre ellas, haciendo un solo gráfico de barras donde aparezcan ambos indicadores.
- Recuerda agregar título y etiquetas a los ejes de tus gráficos.
- Observa y comenta los patrones, valores extremos o posibles relaciones que identifiques.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de estilo
sns.set_theme(style='whitegrid')

plt.figure(figsize=(8, 5))

# Crear boxplot mostrando la media con showmeans=True
sns.boxplot(
    data=merged, 
    x='jams_delay', 
    color='skyblue', 
    showmeans=True,
    meanprops={"marker": "o", "markerfacecolor": "red", "markeredgecolor": "red", "markersize": "7"}
)

# Obtener promedio para mostrarlo en el título
mean_value = merged['jams_delay'].mean()
plt.title(f'Boxplot de JamsDelay (2024)\nPromedio: {mean_value:.2f}', fontsize=12, fontweight='bold')
plt.xlabel('Minutos de retraso (Jams Delay)', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

# Histograma con curva de densidad KDE
sns.histplot(data=merged, x='city_gdp_capita', kde=True, color='teal', bins=10)

# Línea vertical del promedio
mean_gdp = merged['city_gdp_capita'].mean()
plt.axvline(mean_gdp, color='red', linestyle='--', linewidth=2, label=f'Promedio: {mean_gdp:,.0f}')

plt.title('Distribución del PIB per Cápita por Ciudad (2024)', fontsize=12, fontweight='bold')
plt.xlabel('PIB per Cápita (USD)', fontsize=10)
plt.ylabel('Frecuencia (N° de Ciudades)', fontsize=10)
plt.legend()

plt.tight_layout()
plt.show()

```

```python
# Ordenar ciudades para facilitar la lectura visual
df_plot = merged.sort_values(by='city_gdp_capita', ascending=False)

fig, ax1 = plt.subplots(figsize=(12, 6))

width = 0.4
x_indices = range(len(df_plot))

# Barra 1: PIB per Cápita (Eje Y izquierdo)
bars1 = ax1.bar([x - width/2 for x in x_indices], df_plot['city_gdp_capita'], width=width, color='steelblue', label='PIB per Cápita')
ax1.set_ylabel('PIB per Cápita (USD)', color='steelblue', fontsize=11, fontweight='bold')
ax1.tick_params(axis='y', labelcolor='steelblue')
ax1.set_xlabel('Ciudad', fontsize=11)
ax1.set_xticks(x_indices)
ax1.set_xticklabels(df_plot['city'])

# Barra 2: Jams Delay (Eje Y derecho)
ax2 = ax1.twinx()
bars2 = ax2.bar([x + width/2 for x in x_indices], df_plot['jams_delay'], width=width, color='coral', label='Jams Delay (min)')
ax2.set_ylabel('Minutos de Retraso (Jams Delay)', color='coral', fontsize=11, fontweight='bold')
ax2.tick_params(axis='y', labelcolor='coral')

plt.title('Comparativa por Ciudad: PIB per Cápita vs Retraso por Congestión (2024)', fontsize=13, fontweight='bold')
plt.grid(False)

# Rotar etiquetas del eje X a 90 grados antes de mostrar
ax1.tick_params(axis='x', labelrotation=90)

plt.tight_layout()
plt.show()
```

# 🧾 Resumen ejecutivo

Este estudio evaluó la relación entre la movilidad urbana y la productividad económica para determinar si un mayor ingreso mitiga la congestión vial. El análisis consolidó datos de 15 ciudades de 7 países para el año 2024, contrastando métricas clave de congestión (jams_delay, traffic_index_live, tiempos de viaje) contra el PIB per cápita (city_gdp_capita), desempleo y población.

Metodología:

Se procesaron los datasets estandarizando nombres a snake_case, convirtiendo fechas a tipo datetime y limpiando cadenas numéricas a float. Los datos de tráfico se agregaron a promedios por ciudad-año y se integraron con la información económica mediante un inner join con las claves city y year. Se utilizaron boxplots e histogramas para validar la asimetría de los retrasos y la dispersión del ingreso.

Hallazgos Clave:

Los resultados evidencian que no existe una relación lineal directa entre PIB per cápita y congestión:

Falta de correlación directa: Ciudades de alto ingreso como Montevideo (26,176 USD) y Brasilia (16,251 USD) presentan retrasos mínimos (50.2 y 101.6 min respectivamente), mientras que Ciudad de México representa el outlier más crítico con 2,833.1 minutos de retraso pese a contar con un PIB per cápita elevado (21,111 USD).

Peso de la escala urbana: Megaciudades como São Paulo (1,729.2 min de retraso) y Ciudad de México confirman que la densidad poblacional y la morfología urbana impactan el tráfico mucho más que el ingreso per cápita.

Ciudad Prioritaria & Recomendaciones:

Al comparar las tres urbes foco, Bogotá es la ciudad prioritaria para inversión estratégica:

Bogotá combina el nivel de congestión más severo de las tres (1,141.6 min de retraso, índice de tráfico de 37.61 y 140.9 km de atascos promedio) con el menor PIB per cápita del grupo (11,442 USD) y 10.0% de desempleo.

Lima reporta una congestión similar (1,052.3 min), pero con mayor PIB per cápita (13,472 USD) y menor desempleo (6.5%).

Buenos Aires, a pesar de sus 15.4 millones de habitantes, sufre menos de la mitad del retraso (571.1 min), respaldada por un PIB de 18,117 USD y una red de transporte guiado más robusta.

Se recomienda concentrar el financiamiento en Bogotá hacia infraestructura de transporte masivo segregado (líneas de metro y trenes de cercanías) para frenar las pérdidas en productividad, así como incorporar en etapas posteriores variables de densidad vial y transporte público en un modelo multivariable.
